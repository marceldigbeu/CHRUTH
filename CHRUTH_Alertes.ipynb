# CHRUTH - Alertes appels d'offres

Interface pour `ao_alertes.py`. Envoie par email les **nouveaux AO CHAUD/TIEDE**
(avec le brouillon de message), piloté par le drapeau notifications ON/OFF.

Secrets SMTP : `alertes_secrets.json` (ou variables d'environnement). Destinataires : `destinataires.txt`.

Par défaut ce notebook est en **aperçu** : rien n'est envoyé tant que `ENVOYER=False`.

## 1. Setup

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd()))
from datetime import datetime
import ao_alertes, ao_config
print('Modules charges.')
```

## 2. État actuel

```python
print('Notifications actives :', ao_config.notifications_actives())
print('Destinataires        :', ao_alertes.charger_destinataires() or '(aucun)')
en_attente = ao_alertes.nouveaux_ao_a_alerter()
print('Nouveaux AO a alerter :', len(en_attente))
```

## 3. Aperçu de l'email (sans envoi)

Construit l'email des AO en attente et l'écrit dans `output/apercu_alerte.html` (ouvrir dans un navigateur). N'envoie rien, ne marque rien.

```python
en_attente = ao_alertes.nouveaux_ao_a_alerter()
if not en_attente:
    print('Aucun nouvel AO CHAUD/TIEDE a alerter pour le moment.')
else:
    sujet, html, texte = ao_alertes.construire_email(en_attente, datetime.now())
    chemin = pathlib.Path('output') / 'apercu_alerte.html'
    chemin.parent.mkdir(parents=True, exist_ok=True)
    chemin.write_text(html, encoding='utf-8')
    print('Sujet   :', sujet)
    print('Apercu HTML ecrit :', chemin)
    print('\n' + texte)
```

## 4. Envoyer réellement (opt-in)

Mets `ENVOYER = True` pour envoyer vraiment. Respecte le drapeau notifications (rien n'est envoyé si OFF) et marque les AO envoyés.

```python
ENVOYER = False   # passe a True pour envoyer pour de vrai
if ENVOYER:
    n = ao_alertes.envoyer_alertes()
    print(n, 'AO envoyes.' if n else 'aucun nouvel AO (ou notifications OFF).')
else:
    print('Mode apercu : ENVOYER=False, aucun email envoye.')
```
