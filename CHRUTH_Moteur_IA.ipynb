## 2. Diagnostic du moteur

```python
print('Moteur auto     :', llm_client.moteur_auto() or 'aucun -> repli deterministe')
print('Fournisseur cloud:', llm_client.cloud_provider() or '(aucune cle cloud)')
print('Ollama dispo    :', llm_client.llm_disponible('ollama'))
cles = {'ANTHROPIC_API_KEY', 'MISTRAL_API_KEY', 'GROQ_API_KEY'}
presentes = [k for k in cles if os.environ.get(k)]
print('Cles cloud presentes :', presentes or '(aucune)')
```

## 3. Tester un prompt

Édite `SYSTEM` et `PROMPT`. Utilise le moteur détecté. Si aucun moteur (pas de clé, Ollama éteint), rien n'est appelé.

```python
SYSTEM = 'Tu es un assistant concis. Reponds en francais.'
PROMPT = 'Redige une phrase de presentation pour une societe de nettoyage en Ile-de-France.'

provider = llm_client.moteur_auto()
if provider is None:
    print('Aucun moteur : colle une cle cloud dans .env ou lance Ollama.')
else:
    print('Moteur utilise :', provider)
    print('-' * 60)
    print(llm_client.generer(PROMPT, SYSTEM, provider=provider))
```
