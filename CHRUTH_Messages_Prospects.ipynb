# CHRUTH - Messages prospects (par segment)

Interface pour `prospect_messages.py`. Génère un **email + un script d'appel**
pour un **segment** (catégorie × priorité), rendu sur un prospect exemple.

La **fiche CHRUTH** (`config_chruth/fiche_chruth.md`) et le **moteur auto**
(clé cloud / Ollama / repli) rendent le message précis. Brouillons seulement.

## 1. Setup

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd()))
import prospect_messages as pm, llm_client
print('Moteur IA   :', llm_client.moteur_auto() or 'aucun -> brouillon deterministe')
print('Fiche CHRUTH:', 'chargee' if pm.fiche_chruth() else 'VIDE (remplis config_chruth/fiche_chruth.md)')
```

## 2. Choisir le segment et un prospect exemple

Catégories : ex. `PRIV_BUREAU`, `SANTE`, `COMMERCE`… Priorités : `CHAUDE` / `TIEDE`.

```python
CATEGORIE = 'PRIV_BUREAU'
PRIORITE = 'CHAUDE'

DENOMINATION = 'CABINET DENTAIRE DU MARAIS'
VILLE = 'PARIS 03'
EFFECTIF = '10 a 19'
```

## 3. Générer le message

```python
templates = pm.generer_templates([(CATEGORIE, PRIORITE)], refresh=True)
tpl = templates[f'{CATEGORIE}|{PRIORITE}']
ligne = {'denomination': DENOMINATION, 'libelle_commune': VILLE, 'effectif_label': EFFECTIF}
print('Source :', tpl.get('source', ''), '(ia = redige par le modele ; defaut = brouillon type)')
print('\n' + '=' * 60 + '\nEMAIL\n' + '=' * 60)
print(pm.rendre(tpl['email'], ligne))
print('\n' + '=' * 60 + "\nSCRIPT D'APPEL\n" + '=' * 60)
print(pm.rendre(tpl['script'], ligne))
```
